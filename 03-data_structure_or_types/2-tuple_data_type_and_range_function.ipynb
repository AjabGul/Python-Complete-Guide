## range():
- function returns a sequence of numbers, starting from 0 by default, and increments by 1 (by default), and stops before a specified number.

- range(start, stop, step)

- start	Optional. An integer number specifying at which position to start. Default is 0
- stop	Required. An integer number specifying at which position to stop (not included).
- step	Optional. An integer number specifying the incrementation. Default is 1

```python
# create a list of 10 int with range method
num = list(range(0,10))
num
```

```python
# odd list from 1 to 10
odd = list(range(0,10,3))
odd
```

```python
# even list from 1,20
even = list(range(0,20,2))
even
```

```python
# retrive data from range list
even.index(6)
```

```python
even[3]
```
