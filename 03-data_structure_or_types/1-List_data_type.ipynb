## List
- is a variable that can have a sequence of values assigned to it. In a list, these values are known as elements.
- A list is an ordered and mutable collection in Python that can store multiple items of different data types and is represented using square brackets [].

```python
#  define  List
my_list = ["ajab", "pishin", 35, ["biotehnology", "biochemistry", "bioinformatics"], 5.11 ]
print(my_list)
print(type(my_list))
```

```python
#  define List
cities = list(("pishin", "yaro", "quetta", "chaman"))
print(cities)
print(type(cities))
```

#### List: adding and changing elements in a list

```python
# append the element to the end of the list
cities.append("Kuchlakh")
print(cities)
```

```python
# using + sign to add new list to existing list
cities = cities + ["karachi", "lahor"]
print(cities)
```

```python
# insert method to insert the element to a specfic position
cities.insert(3, "islamabad")
print(cities)
```

```python
# changinglist: add new element to a specific index list[index] = new_element
cities[3] = "multan"
print(cities)
```

#### List: Taking slices out of the list

```python
# indexing
smaller_list_of_cities = cities[2:5]
print(smaller_list_of_cities)
```

```python
print(my_list)
print(f"print list with in the my_list at index 3: \n {my_list[3]}")
```

```python
# print a complete list
my_list[:]
```

#### List: Deleting and removing elements from the list

```python
# del list_name[index] by index position in the list
del my_list[3]
print(my_list)
```

```python
# list_name.remove("element") specifing the element in the list
my_list.remove(5.11)
print(my_list)
```

#### List: popping elements from the list

```python
#  pop elemet from a specific position on default 
my_pop_element = my_list.pop(1)
print(my_pop_element)
print(my_list)
```

#### List: extend, sort, min, max, len and mean 

```python
#  extend method
list_1 = ["a","b","c"]
list_2 = ["k", "l", "d","e","f"]
# extend list 1 by list 2
list_1.extend(list_2)
print(list_1)
```

```python
#  sort()
list_1.sort()
print(list_1)
# sort(resvers = True) in reverse order
list_1.sort(reverse=True)
print(list_1)
```

```python
score = [23,45,34,22,5,77,8,99]
# min
print(f"min element of the list score = {min(score)}")
# max
print(max(score))
# sum
print(sum(score))
# len of the list
print(len(score))
# mean
print(sum(score)/len(score))
```
