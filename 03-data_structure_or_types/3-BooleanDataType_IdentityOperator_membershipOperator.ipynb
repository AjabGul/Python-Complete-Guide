## Boolean Data Type

- Booleans represent one of two values: True or False

- Boolean Values

- In programming you often need to know if an expression is True or False.

- You can evaluate any expression in Python, and get one of two answers, True or False.

- When you compare two values, the expression is evaluated and Python returns the Boolean answer:

```python
a = True
b = False
print(a)
print(b)
print(type(a))
print(type(b))
```

```python
# comparison opreators with bolean
print(10 > 9)
print(10 == 9)
print(10 < 9)
print(10 != 9)
```

```python
#  logical operators with boolean data type

#  & logical operator both condition must be true
print(True & True)
print(True & False)
#  or logical operator one must be True
print(True or True)
print(True or False)
```

```python
c = 3
d = 5

# & logical operator
print(c <= d & d >= c)
print(c == d & d >= c)
#  0r ligical operator
print(c <= d or d == c)
print(c == d or d == c)
```

####  Identity Operator:
- is used to check whether two variables refer to the same object in memory, not merely whether their values are equal.

- Identity operators

        -- Operator                    Meaning

        -- is                  Returns True if both variables refer to the same object

        --is not               Returns True if they refer to different objects

```python
a = [1, 2, 3]
b = a
c = [1, 2, 3]

print(a is b)       # True
print(a is c)       # False
print(a == c)       # True
print(a is not c)   # True
```

#### in and in not Membership Operator:

- it is used to check whether a value exists inside a sequence such as a list, tuple, string, or set.

- Membership operators

- Python has two membership operators:

        - Operator	           Meaning
        --in	               Checks if a value exists
        --not in	           Checks if a value does not exist

```python
# memebership operator with list
names = ["ajab", "gul", "kakar", "bioinformatics", "molecular biology"]
print("ajab" in names)
print("ajab" not in names)
print("math" in names)
print("math" not in names)
```

```python
# string
bio = """
Dear Ajab,

it is kindly requested to upload your bio data in online application form!

"""

print("bio" in bio)
print("bio" not in bio)
print("address" in bio)
print("address" not in bio)
```

```python
genes = ["rpoB", "katG", "gyrA", "BRCA1", "TP53", "EGFR"]

if "rpoB" in genes:
    print(f" rpoB gene is present in genes")

print("rpoB" in genes)
print("rpoB" not in genes)
```
