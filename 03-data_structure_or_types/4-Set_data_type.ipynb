## Set
- a set is a collection that stores unique values. A set is written using curly brackets {}.
- Sets automatically remove duplicates

```python
# define set
A = {'A', 'B', 'B', 'D', 'A'}
B = {'X', 'Y', 'C', 'X', 'B'}

print(A)
print(B)

```

```python
#  intersection and union of the set
C = A.intersection(B)
D = A.union(B)

print(C)
print(D)

```

```python
#  add and update the set
A.add('John cena')
A.update(['John2'])


print(len(A))
print(A)
```

```python
# reomove the items form set
A.remove('John2')
A.remove('John cena')

print(A)
```

```python
# romove the complete set
del A
```
