## Dictionary Data Type
- Data is stored as key : value pair
- unorder and mutable data type
- Keys must be unique
- Values can be duplicated
- Values can have different data types

```python
student = {
    "name": "Ajab",
    "age": 37,
    "masters": ("Biotecnology and Informatics", "Biocemistry and Molecular Biology"),
    "field": "Biotechnology",
    'university': 'BUITEMS',
    "research" : ["genomics", "bioinformatics", "genetics"]
}

print(student)
```

```python
# get the key and values
print(student.keys())
print(student.values())
```

```python
# get the value by key
student["name"]
```

```python
#  if try to get the value of a key which is not exist will get error
# but with get method we will get none instead of error
print(student.get("kakar"))
```

#### retriving values from dict

```python
data = {
    "name": "Ajab",
    "age": 35,                    # integer
    "weight": 75.5,               # float
    "student": True,              # boolean
    "skills": ["PCR", "Python"],  # list
    "address": {                   # another dictionary
        "city": "Quetta",
        "country": "Pakistan"
    }
}
print(data)
```

```python
# retrive data usnig key
print(data["name"])
print(data["age"])
print(data["skills"])
print(data["skills"][1])
print(data["address"])
print(data["address"]["city"])
print(data["address"]["country"])
```

```python
data
```

```python
# access all items of the dict
print(data.items())
```

```python
# items() method retun tuple now how to unpack the tuple
# we can also unpack data.keys() and data.values 
a,b,c,d,e,f,g = data.items()
print(a)
print(b)
print(c)
print(d)
print(e)
print(f)
print(g)
```

```python
# change key values
data["weight"] = 85
# add new key values
data["education"] = " Master in Biotechnology and Informatics"
```

```python
print(data)
```

```python
# convert the tuple in dict. but tuple must be in pairs
my_tuple = (
    ("name", "Ajab"),
    ("age", 35),
    ("city", "Quetta")
)

new_dict = dict(my_tuple)

print(new_dict)
```

```python
# pop list items of dict
new_dict.popitem()
```
