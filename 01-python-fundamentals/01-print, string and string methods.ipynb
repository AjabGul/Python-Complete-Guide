## 3- Method used for string

```python
sub = "   Molecular Biology.txt and informtics.txt     "
```

```python

#  Upper case
print(sub.upper())

# lower case
print(sub.lower())


# title(): capitalize each word
print(sub.title())

# strrp(): remove white space from both ends"
print(sub.strip())
# lstrip(): remove space from left side
print(sub.lstrip())
# rstrip() remove space from right side
print(sub.rstrip())

# replace(): replace text
print(sub.replace("Biology", "Biologist"))


# split(): string into list
print(sub.split())
# split where .
print(sub.split("."))
# split from the right side
print(sub.rsplit(".", 1))


print(len(sub))
```

```python
# find(): find char position in the string
subject = ("Moleuclar Bilogy")


print(subject.find("B"))
# find from the right
print(subject.rfind("o"))


# index find the position of the text
subject.index("B")


# count() count the chr in a string
subject.count("l")
```

```python
# add - one the right
print(subject.rjust(30, "-"))
# add - on the left
print(subject.ljust(30, "-"))

# zfill() fill zero's befor and string numb
score = "77"
print(score.zfill(6))
```

#### Then learn the checking methods:

```python
# the following method return false or true

print(subject.isalpha())
print(subject.isdigit())
print(subject.isalnum())
print(subject.isspace())
print(subject.islower())
print(subject.isupper())
print(subject.istitle())


```
